```python
import cv2 as cv 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
img = cv.imread("image1.png")
cv.imshow("Image", img)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
gray_img = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
cv.imshow("Gray Image", gray_img)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
## Create Prewit Filter

kernelx = np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
kernely = np.array([[1,1,1],[0,0,0],[-1,-1,-1]])

img_prewitx = cv.filter2D(img, -1, kernelx)
img_prewity = cv.filter2D(img, -1, kernely)

prewitt = np.hypot(img_prewitx, img_prewity)
prewitt = prewitt[:,:,0]
prewitt = prewitt.astype("int")
plt.imshow(prewitt, cmap="gray")
```

```python
## Sobel FIlter 

kernelx = np.array([
    [-1,0,1],
    [-2,0,2],
    [-1,0,1]
])
kernely = np.array([
    [1,2,1],
    [0,0,0],
    [-1,-2,-1]
])
gray_img_sobelx = cv.filter2D(gray_img, -1, kernelx)
gray_img_sobely = cv.filter2D(gray_img, -1, kernely)
new = np.hypot(gray_img_sobelx, gray_img_sobely)
plt.imshow(new.astype("int"), cmap = "gray")
```

```python
# Canny Edge Detection 

image = img 

(H,W) = image.shape[:2]

gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
blurred = cv.GaussianBlur(gray, (5,5),0)
canny = cv.Canny(blurred, 30, 150)
fig, ax = plt.subplots(1,2, figsize = (18,18))
ax[0].imshow(gray, cmap = "gray")
ax[1].imshow(canny, cmap = "gray")

ax[0].axis("off")
ax[1].axis("off")
```
