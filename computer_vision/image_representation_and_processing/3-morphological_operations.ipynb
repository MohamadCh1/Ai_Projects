```python
import cv2 as cv 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
def load_image():
    blank_image = np.zeros((600,600))
    cv.putText(blank_image, text = 'ABCDE', org = (50,300), fontFace = cv.FONT_HERSHEY_SIMPLEX, fontScale = 5,
               color = (255,255,255), thickness = 25, lineType = cv.LINE_AA)
    return blank_image 
```

```python
# def display_image(window_name, img):
#     cv.imshow(window_name, img)
#     if cv.waitKey(0):
#         cv.destroyAllWindows()

def display_image(img):
    fig = plt.figure(figsize=(12,10))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap = 'gray')
```

```python
img = load_image()
display_image(img)
```

### Erosin

```python
kernel = np.ones((5,5), np.uint8)
erosin1 = cv.erode(img, kernel, iterations = 1)
display_image(erosin1)
```

### Dilation 

```python
kernel = np.ones((5,5), np.uint8)
dialation1 = cv.dilate(img, kernel, iterations = 1)
display_image(dialation1)
```

### Opening 

```python
## Opening = Erosion + Dialation
img = load_image()
white_noise = np.random.randint(0,2,(600,600))
white_noise
```

```python
white_noise = white_noise *255
display_image(white_noise)
```

```python
noise_img = white_noise + img
display_image(noise_img)
```

```python
opening = cv.morphologyEx(noise_img, cv.MORPH_OPEN, kernel)
display_image(opening)
```

```python
img = load_image()
black_noise = np.random.randint(0,2,(600,600)) * -255
black_noise_image = black_noise + img
display_image(black_noise_image)
```

```python
black_noise_image[black_noise_image==-255]=0
display_image(black_noise_image)
```

```python
closing = cv.morphologyEx(black_noise_image, cv.MORPH_CLOSE, kernel)
display_image(closing)
```

```python
gradient = cv.morphologyEx(img, cv.MORPH_GRADIENT, kernel)
display_image(gradient)
```
