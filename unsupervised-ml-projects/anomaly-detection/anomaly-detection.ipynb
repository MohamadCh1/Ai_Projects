```python
from sklearn.preprocessing import MinMaxScaler
X = data.drop("Class", axis=1)
scalar = MinMaxScaler()
X_scaled = pd.DataFrame(scalar.fit_transform(X), columns=X.columns)
X_scaled.head()
```

```python
from sklearn.decomposition import PCA

# Reduce to 2 components for visualization
pca = PCA()
pca_df = pd.DataFrame(pca.fit_transform(X_scaled))

plt.step(range(1,len(pca.explained_variance_ratio_)+1), np.cumsum(pca.explained_variance_ratio_), where='mid')
plt.plot(np.cumsum(pca.explained_variance_ratio_),marker = 'o')
plt.xlabel("Number of principle componenet")
plt.ylabel("Cumulative Explained Varience Ratio")
plt.title("Varience covered by principle component")
plt.grid(True)
plt.show()
```

```python
optimal_component = np.argmax(np.cumsum(pca.explained_variance_ratio_)>=0.95)+1
print(f"Optimal number of Component: {optimal_component} ")
```

🧠 Interpretation
- Recall is the most critical metric here — it tells us how many actual frauds were successfully detected.
- Isolation Forest detects ~18% of frauds.
- One-Class SVM detects ~13% of frauds.
- Precision reflects how many of the flagged frauds were actually fraudulent.
- Isolation Forest has higher precision than One-Class SVM, but both are relatively low.
- F1-score, the harmonic mean of precision and recall, is also higher for Isolation Forest, indicating a better balance between catching frauds and minimizing false alarms.

✅ Conclusion
While both models achieve high overall accuracy (>99%), this is expected due to the overwhelming majority of normal transactions. When focusing on the rare fraud class:
- Isolation Forest outperforms One-Class SVM across all key metrics.
- However, recall remains low, meaning most fraud cases are still missed.
