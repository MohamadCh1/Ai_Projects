```python
(R,G,B) = cv.split(img) 
```

```python
cv.namedWindow("Blue", cv.WINDOW_NORMAL)
cv.namedWindow("Green", cv.WINDOW_NORMAL)
cv.namedWindow("Red", cv.WINDOW_NORMAL)
```

```python
cv.imshow("Blue",B)
cv.imshow("Green",G)
cv.imshow("Red",R)
```

```python
cv.imwrite("Channel_red.jpg",R)
cv.imwrite("Channel_green.jpg",G)
cv.imwrite("Channel_blue.jpg",B)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

### Drawing

```python
canvas = np.zeros((300,300,3), dtype = "uint8")
```

```python
green = (0,255,0)
cv.line(canvas, (0,0), (300,300), green)
cv.imshow("canvas", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
red = (0,0,255)
cv.line(canvas, (300,0), (0,300), red)
cv.imshow("canvas", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
blue = (255,0,0)
cv.rectangle(canvas, (100,10), (210,60), blue, 2)
cv.imshow("canvas", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
blue = (100,0,0)
cv.rectangle(canvas, (100,10), (210,60), blue, -1)
cv.imshow("canvas", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
canvas = np.zeros((300,300,3), dtype = "uint8")
(centerX, centerY) = (canvas.shape[1] // 2, canvas.shape[0] // 2)
white = (255,255,255)
```

```python
for r in range(0,150,25):
    cv.circle(canvas, (centerX, centerY), r, white)
cv.imshow("canvas", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```

```python
for i in range(20):
    radius = np.random.randint(5, high=200)
    color = np.random.randint(0, high=200, size=(3,)).tolist()
    pt = np.random.randint(0, high = 300, size = (2,))
    cv.circle(canvas, pt, radius, color, -1)
cv.imshow("canvas", canvas)
if cv.waitKey(0):
    cv.destroyAllWindows()
```
